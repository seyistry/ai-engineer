# Self- Consistency

````python

persona = "Imagine three completely independent expert who reason differently are answering a question. The final answer is obtained by majority vote."


# A bookstore has 20 books on a shelf. A customer buys 5 books. Then the store restocks with double the number of books bought. After, another customer buys 8 books. How many books are left on the shelf?

userInput = input("text: ")
final_input=f"{persona}\n\n {userInput}"

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents= final_input
)


print(f"{response.text}")
````
