# Prompting

````python
import ollama
````

## Zero shot

````python
messages = [
    {'role': 'user', 'content': 'Translate this to French: "Today is a good day"'}
]

response = ollama.chat(model='gemma3:1b', messages=messages)
print(response['message']['content'])
````

## One-shot Prompting

````python
messages = [
    {'role': 'user', 'content': """
  									Detect the language of the following sentences:
									text: "A plus tard" -> language: "French"
									Now detect:
									text: "Gracias!"
  								"""}
]

response = ollama.chat(model='gemma3:1b', messages=messages)
print(response['message']['content'])
````

## Few-Shot Prompting

````python
messages = [
    {'role': 'user', 'content': """
  									Determine the sentiment of the following sentences:
									text: "I love this product! it works perfectly." -> Classification: "Positive"
									text: "The service was terrible. i'm very disappointed." -> Classification: "Negative"
									text: "The food was okay, nothing special." -> Classification: "Neutral"
									Now analyze this sentence:
									text: "The movie was amazing, I enjoyed every moment of it!"
  								"""}
]

response = ollama.chat(model='gemma3:1b', messages=messages)
print(response['message']['content'])
````

## Multi-step Prompting

````python
text = "Marie Curie discovered radium 1898"
messages = [
    {'role': 'user', 'content': f"""
  									You will be provided with a text delimited by triple backticks,
									Step 1: Identify the key entities in the sentence.
									Step 2: Classify each entity as a person, event, or object.
									```{text}```
  								"""}
]

response = ollama.chat(model='gemma3:1b', messages=messages)
print(response['message']['content'])
````

## Chain-of-thought Prompting

````python
messages = [
    {'role': 'user', 'content': f"""
  									John has 3 apples, he buys 5 more and gives away 2, How many apples does he have now.
									Think step by step.
  								"""}
]

response = ollama.chat(model='gemma3:1b', messages=messages)
print(response['message']['content'])
````

## Self - Consistency Prompting

````python

instruction = "Imagine three completely independent expert who reason differently are answering a question. The final answer is obtained by majority vote. The question is: "


question = "A bookstore has 20 books on a shelf. A customer buys 5 books. Then the store restocks with double the number of books bought. After, another customer buys 8 books. How many books are left on the shelf?"

prompt = instruction + question

messages = [
    {'role': 'user', 'content': prompt}
]

response = ollama.chat(model='qwen3:1.7b', messages=messages)
print(response['message']['content'])
````
